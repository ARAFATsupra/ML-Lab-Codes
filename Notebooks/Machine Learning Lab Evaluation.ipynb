4 How many categorical columns?

```python
categorical_cols = titanic.select_dtypes(include=['object', 'category']).columns
print(f"Number of categorical columns: {len(categorical_cols)}")
```

5 One hot encoding

```python
import pandas as pd

cols_to_encode = ['Sex', 'Embarked']
one_hot = pd.get_dummies(titanic[cols_to_encode])
one_hot.head()
```

```python
titanic = pd.concat([titanic, one_hot], axis=1)
titanic[30:50]
```

6 Handling missing values, replace with median

```python
# Fill missing numeric values with median
titanic = titanic.fillna(titanic.median(numeric_only=True))

cols_to_drop = ['Name', 'Ticket', 'Cabin', 'Sex', 'Embarked']
titanic = titanic.drop(columns=[c for c in cols_to_drop if c in titanic.columns])

print("Missing values per column:")
print(titanic.isnull().sum())
```

7 Separate input, output

```python
# Separate features (X) and target (y)
y = titanic["Survived"]
X = titanic.drop("Survived", axis=1)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

8 Split into 70:30 training & test set

```python
# Step 7: Split the dataset into training and test sets using the cleaned X and y
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
```

  9 Model Train

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score, roc_auc_score
```

```python
# Initialize the KNN classifier
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
# Train the KNN model
knn.fit(X_train, y_train)
```

```python
# Make predictions on the test set
y_pred = knn.predict(X_test)
```

10 Save the model in your drive

```python
import joblib

# Save model to Google Drive
joblib.dump(knn, '/content/drive/MyDrive/knn_model.pkl')
```

11 Evaluate the model

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)

print("Model Performance:")
print(f"R2 Score: {r2:.4f}")
```
