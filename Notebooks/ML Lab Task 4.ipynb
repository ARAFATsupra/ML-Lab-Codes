```python
# Load the MNIST dataset (from tensorflow)
import tensorflow as tf
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Number of samples to display
num_samples = 5

# Randomly select 'num_samples' images from the training set
indices = np.random.choice(x_train.shape[0], num_samples, replace=False)

# Create a figure to display the images
fig, axes = plt.subplots(1, num_samples, figsize=(15, 5))

for i, ax in zip(indices, axes):
    ax.imshow(x_train[i], cmap='gray')  # Display the image
    ax.set_title(f"Label: {y_train[i]}")  # Display the label
    ax.axis('off')  # Hide the axis

plt.tight_layout()
plt.show()
```

```python
# Normalize pixel values to be between 0 and 1
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
# Flatten the images for KNN (from 28x28 to 784,)
x_train_flattened = x_train.reshape(-1, 28 * 28)
x_test_flattened = x_test.reshape(-1, 28 * 28)
```

**train and Test a KNN model**

```python
#import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score, roc_auc_score
```

```python
# Initialize the KNN classifier (you can experiment with n_neighbors)
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
# Train the KNN model
knn.fit(x_train_flattened, y_train)
```

```python
# Make predictions on the test set
y_pred = knn.predict(x_test_flattened)
```

```python
# 1. Accuracy
accuracy = accuracy_score(y_test, y_pred)

# 2. Precision
precision = precision_score(y_test, y_pred, average='weighted')

# 3. Recall
recall = recall_score(y_test, y_pred, average='weighted')

# 4. F1 Score
f1 = f1_score(y_test, y_pred, average='weighted')
```

```python
# Print all metrics
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
```

**Train and test Logistic regression Classifier**

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=2000)
# Train the Logistic Regression model
log_reg.fit(x_train_flattened, y_train)
# Make predictions on the test set
y_pred_log = log_reg.predict(x_test_flattened)
# Calculate Precision, Recall, Accuracy, F1 Score, and AUC
# 1. Accuracy
accuracy = accuracy_score(y_test, y_pred_log)
# 2. Precision
precision = precision_score(y_test, y_pred_log, average='weighted')
# 3. Recall
recall = recall_score(y_test, y_pred_log, average='weighted')
# 4. F1 Score
f1 = f1_score(y_test, y_pred_log, average='weighted')


# Print all metrics
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
```

**train and test Decision Tree Classifier**

```python
from sklearn.tree import DecisionTreeClassifier
# --- Gini Decision Tree ---
dt_gini = DecisionTreeClassifier(criterion='gini', random_state=42)
dt_gini.fit(x_train_flattened, y_train)

y_pred_gini = dt_gini.predict(x_test_flattened)

print("Decision Tree (Gini)")
print("Accuracy:", accuracy_score(y_test, y_pred_gini))
print("Precision:", precision_score(y_test, y_pred_gini,average='weighted'))
print("Recall:", recall_score(y_test, y_pred_gini,average='weighted'))
print("F1 Score:", f1_score(y_test, y_pred_gini,average='weighted'))
print("\n")
```

```python
# --- Entropy Decision Tree ---
dt_entropy = DecisionTreeClassifier(criterion='entropy', random_state=42)
dt_entropy.fit(x_train_flattened, y_train)

y_pred_entropy = dt_entropy.predict(x_test_flattened)

print("Decision Tree (Entropy)")
print("Accuracy:", accuracy_score(y_test, y_pred_entropy))
print("Precision:", precision_score(y_test, y_pred_entropy, average='weighted'))
print("Recall:", recall_score(y_test, y_pred_entropy, average='weighted'))
print("F1 Score:", f1_score(y_test, y_pred_entropy, average='weighted'))
print("\n")
```
